```python
import sqlite3

# Connect to the database file (creates it if it doesn't exist)
conn = sqlite3.connect('note_database.db')

# Create a cursor object
cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS profiles")
cursor.execute("DROP TABLE IF EXISTS notes")
cursor.execute("DROP TABLE IF EXISTS sessions")

# Execute a SQL command to create a table
cursor.execute('''
CREATE TABLE IF NOT EXISTS profiles (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    photo_url TEXT,
    history TEXT
)
''')

# Create sessions table
cursor.execute('''
CREATE TABLE IF NOT EXISTS sessions (
    id INTEGER PRIMARY KEY,
    session_name TEXT NOT NULL,
    created_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
''')

# Create notes table
cursor.execute('''
CREATE TABLE IF NOT EXISTS notes (
    id INTEGER PRIMARY KEY,
    session_id INTEGER NOT NULL,
    note_text TEXT NOT NULL,
    created_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (session_id) REFERENCES sessions(id)
)
''')

# Insert some sample data (optional, but shows functionality)
cursor.execute("INSERT INTO profiles (name, age, photo_url, history) VALUES (?, ?, ?, ?)", ('Warlock', 100, "./assets/images/profiles/Warlock.png", "He's really dumb"))

# Commit the changes
conn.commit()

# Close the connection
conn.close()
```
